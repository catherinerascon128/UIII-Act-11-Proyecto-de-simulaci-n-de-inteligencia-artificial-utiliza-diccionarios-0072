```python
# ============================================
# SIMULADOR IA DE PANADEROS
# Lista de diccionarios + funciones + menú
# ============================================

# Lista vacía donde se guardarán los panaderos
panaderos = []

# ------------------------------------------------
# FUNCIÓN: AGREGAR PANADERO
# ------------------------------------------------
def agregar_panadero():

    print("\n=== AGREGAR PANADERO ===")

    tipos_pan = input("Tipos de pan: ")
    horno = input("Horno utilizado: ")
    produccion = input("Producción diaria: ")
    panaderia = input("Nombre de la panadería: ")
    receta = input("Receta secreta: ")
    horario = input("Horario de madrugada: ")

    # Diccionario
    panadero = {
        "Tipos de pan": tipos_pan,
        "Horno utilizado": horno,
        "Producción diaria": produccion,
        "Panadería": panaderia,
        "Receta secreta": receta,
        "Horario de madrugada": horario
    }

    # Guardar en la lista
    panaderos.append(panadero)

    print("\n✅ Panadero agregado correctamente")


# ------------------------------------------------
# FUNCIÓN: LISTAR PANADEROS
# ------------------------------------------------
def listar_panaderos():

    print("\n=========== LISTA DE PANADEROS ===========")

    if len(panaderos) == 0:
        print("No hay panaderos registrados")
    else:

        # Encabezados alineados
        print("-" * 120)

        print(
            f"{'N°':<5}"
            f"{'TIPOS DE PAN':<20}"
            f"{'HORNO':<18}"
            f"{'PRODUCCIÓN':<18}"
            f"{'PANADERÍA':<18}"
            f"{'RECETA':<18}"
            f"{'HORARIO':<18}"
        )

        print("-" * 120)

        contador = 1

        # FOR KEY VALUES
        for panadero in panaderos:

            print(
                f"{contador:<5}"
                f"{panadero['Tipos de pan']:<20}"
                f"{panadero['Horno utilizado']:<18}"
                f"{panadero['Producción diaria']:<18}"
                f"{panadero['Panadería']:<18}"
                f"{panadero['Receta secreta']:<18}"
                f"{panadero['Horario de madrugada']:<18}"
            )

            contador += 1

        print("-" * 120)


# ------------------------------------------------
# FUNCIÓN: BUSCAR PANADERO
# ------------------------------------------------
def buscar_panadero():

    print("\n=== BUSCAR PANADERO ===")

    nombre_panaderia = input("Ingrese la panadería a buscar: ")

    encontrado = False

    for panadero in panaderos:

        if panadero["Panadería"] == nombre_panaderia:

            print("\n✅ PANADERO ENCONTRADO")

            # FOR KEY VALUE
            for key, value in panadero.items():
                print(key, ":", value)

            encontrado = True

    if encontrado == False:
        print("❌ No se encontró la panadería")


# ------------------------------------------------
# FUNCIÓN: MODIFICAR PANADERO
# ------------------------------------------------
def modificar_panadero():

    print("\n=== MODIFICAR PANADERO ===")

    nombre_panaderia = input("Ingrese la panadería a modificar: ")

    encontrado = False

    for panadero in panaderos:

        if panadero["Panadería"] == nombre_panaderia:

            print("\nIngrese los nuevos datos")

            panadero["Tipos de pan"] = input("Nuevo tipo de pan: ")
            panadero["Horno utilizado"] = input("Nuevo horno: ")
            panadero["Producción diaria"] = input("Nueva producción: ")
            panadero["Receta secreta"] = input("Nueva receta: ")
            panadero["Horario de madrugada"] = input("Nuevo horario: ")

            print("✅ Datos modificados correctamente")

            encontrado = True

    if encontrado == False:
        print("❌ Panadería no encontrada")


# ------------------------------------------------
# FUNCIÓN: ELIMINAR PANADERO
# ------------------------------------------------
def eliminar_panadero():

    print("\n=== ELIMINAR PANADERO ===")

    nombre_panaderia = input("Ingrese la panadería a eliminar: ")

    encontrado = False

    for panadero in panaderos:

        if panadero["Panadería"] == nombre_panaderia:

            panaderos.remove(panadero)

            print("✅ Panadero eliminado")

            encontrado = True

    if encontrado == False:
        print("❌ Panadería no encontrada")


# ------------------------------------------------
# FUNCIÓN IA PANADERA
# ------------------------------------------------
def inteligencia_panadera():

    print("\n=== IA PANADERA ===")

    tipo = input("¿Qué tipo de pan deseas producir hoy?: ")

    if tipo == "bolillo":
        print("🥖 La IA recomienda usar horno de piedra")

    elif tipo == "conchas":
        print("🍞 La IA recomienda producción nocturna")

    elif tipo == "baguette":
        print("🥖 La IA recomienda fermentación lenta")

    else:
        print("🤖 La IA recomienda revisar la receta secreta")


# ------------------------------------------------
# MENÚ PRINCIPAL
# ------------------------------------------------
while True:

    print("\n===================================")
    print("      SISTEMA IA PANADEROS")
    print("===================================")

    print("1. Agregar panadero")
    print("2. Listar panaderos")
    print("3. Buscar panadero")
    print("4. Modificar panadero")
    print("5. Eliminar panadero")
    print("6. Inteligencia Panadera")
    print("7. Salir")

    opcion = input("\nSeleccione una opción: ")

    if opcion == "1":
        agregar_panadero()

    elif opcion == "2":
        listar_panaderos()

    elif opcion == "3":
        buscar_panadero()

    elif opcion == "4":
        modificar_panadero()

    elif opcion == "5":
        eliminar_panadero()

    elif opcion == "6":
        inteligencia_panadera()

    elif opcion == "7":
        print("\n👋 Gracias por usar el sistema IA Panaderos")
        break

    else:
        print("❌ Opción incorrecta")
```
